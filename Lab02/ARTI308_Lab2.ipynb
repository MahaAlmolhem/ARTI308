## Machine Learning Problem

The aim of this project is to classify websites as legitimate, suspicious, or phishing based on different website characteristics.

This is a classification problem because the target variable contains categories rather than continuous numerical values.

The target variable is `Result`.

The model is expected to learn patterns from the website features and use them to predict the website category.

## Dataset

The dataset used in this project is the Website Phishing dataset from the UCI Machine Learning Repository.

The dataset contains 1353 records and 10 columns.

There are 9 input features that describe different website characteristics, while the `Result` column is the target variable used for classification.

```python
import pandas as pd
from scipy.io import arff
```

```python
data = arff.loadarff("PhishingData.arff")
df = pd.DataFrame(data[0])
```

```python
for column in df.columns:
    df[column] = df[column].str.decode("utf-8").astype(int)
```
